  <p style="color:#b9c8da;font-size:13px;line-height:1.55;margin:0;">Carregue o motor central antes de executar as etapas de partitura seguintes.</p>

```python
#@title 6 - Partitura: PAINEL + motor (rode antes da Celula 7) { display-mode: "form" }
# ============ MOTOR CENTRAL DE PARTITURA v14 (motor v12 INTOCADO dentro) ============
# Desenvolvido por Raphael Silveira.
import numpy as np, librosa, os, re
from scipy.signal import find_peaks
from music21 import stream, note as m21note, tempo as m21tempo, meter, key, duration, metadata, tie as m21tie
from music21 import metadata as m21meta, instrument as m21inst, clef as m21clef
NOTAS=['C','C#','D','D#','E','F','F#','G','G#','A','A#','B']

# ---- detector de BPM (motor v12, IDENTICO) ----
def detectar_bpm_orivox(y, sr):
    if sr>22050:
        f=int(round(sr/22050)); y=y[::f]; sr=sr//f
    N,HOP=1024,512; nF=max(8,(len(y)-N)//HOP); hopS=HOP/sr
    win=0.5-0.5*np.cos(2*np.pi*np.arange(N)/N); odf=np.zeros(nF); prev=np.zeros(N//2)
    for fr in range(nF):
        s=fr*HOP; mag=np.abs(np.fft.rfft(y[s:s+N]*win)[:N//2])
        m=np.log(1+mag); d=m-prev; odf[fr]=np.maximum(0,d).sum(); prev=m
    Wm=max(2,round(0.35/hopS))
    def realca(src):
        out=np.zeros(len(src)); acc=0; fila=[]
        for i in range(len(src)):
            fila.append(src[i]); acc+=src[i]
            if len(fila)>Wm: acc-=fila.pop(0)
            out[i]=max(0,src[i]-acc/len(fila))
        return out
    odf2=realca(odf); meanO=odf2.mean() if nF else 0
    def acorrE(L):
        L=int(round(L))
        if L<2 or L>=nF-2: return 0
        a=odf2[:nF-L]; b=odf2[L:]
        return float((a*b).mean()) if len(a) else 0
    def combE(b):
        L=(60/b)/hopS
        return acorrE(L)+0.7*acorrE(L*2)+0.5*acorrE(L*3)+0.35*acorrE(L*4)
    curve=[(b,combE(b)) for b in np.arange(40,220.5,0.5)]; cands=set()
    for i in range(2,len(curve)-2):
        v=curve[i][1]
        if v>curve[i-1][1] and v>=curve[i+1][1] and v>curve[i-2][1] and v>=curve[i+2][1]: cands.add(curve[i][0])
    if not cands: cands.add(100)
    for b in list(cands):
        for v in [b/2,b*2,b*3/4,b*4/3]:
            if 40<=v<=220: cands.add(round(v*2)/2)
    def beatScore(b):
        P=(60/b)/hopS; best=0; bph=0
        for s in range(24):
            ph=s*P/24; idx=np.arange(ph,nF,P).round().astype(int); idx=idx[idx<nF]
            v=odf2[idx].mean() if len(idx) else 0
            if v>best: best=v; bph=ph
        return best/(meanO+1e-9),bph
    def prefLento(b): return 1+0.12*np.exp(-0.5*((b-90)/60)**2)
    bestB=100; bestV=-1; bestPh=0
    for b in cands:
        sc,ph=beatScore(b); v=sc*prefLento(b)
        if v>bestV: bestV=v; bestB=b; bestPh=ph
    for b in np.arange(bestB-1.5,bestB+1.55,0.05):
        if b<40 or b>220: continue
        sc,ph=beatScore(b); v=sc*prefLento(b)
        if v>bestV: bestV=v; bestB=b; bestPh=ph
    for mult in [1,2,0.5,1.5]:
        cand=bestB*mult
        if 60<=cand<=100: bestB=cand; break
    return round(bestB*100)/100, round(max(0,bestPh*hopS),3)

# ---- CAMADA 1: analise-mestre (roda UMA vez, no audio completo) ----
_ORIVOX_CTX='/content/orivox_contexto.json'
def _salvar_ctx(ctx):
    import json as _j
    try:
        d={k:v for k,v in ctx.items() if k in ('bpm','toni','tipo','fase')}
        _j.dump(d, open(_ORIVOX_CTX,'w'))
    except Exception: pass
def _ler_ctx():
    import json as _j, os as _o
    if _o.path.exists(_ORIVOX_CTX):
        try: return _j.load(open(_ORIVOX_CTX))
        except Exception: return None
    return None

def analisar_musica(caminho_audio_completo, bpm_manual=None, usar_contexto=True):
    # PADRONIZACAO: se ja existe um contexto-mestre salvo (BPM/tom detectados uma
    # vez) e o usuario nao forcou bpm_manual, REUSA - assim todas as celulas usam
    # o MESMO BPM e tom. Coerencia total. (nao muda as NOTAS, so o BPM/tom base.)
    if usar_contexto and not bpm_manual:
        _c=_ler_ctx()
        if _c and all(k in _c for k in ('bpm','toni','tipo')):
            _c=dict(_c)
            graus=[0,2,4,5,7,9,11] if _c['tipo']=='maior' else [0,2,3,5,7,8,10]
            _c['escala']=set((_c['toni']+g)%12 for g in graus)
            _c['batida']=60.0/_c['bpm']; _c.setdefault('fase',0.0)
            return _c
    y,sr=librosa.load(caminho_audio_completo,sr=22050,mono=True)
    bpm_auto,fase=detectar_bpm_orivox(y,sr)
    bpm=float(bpm_manual) if bpm_manual else bpm_auto
    try:
        _m=_get_bpm_mult()
        if _m and _m!=1.0: bpm=bpm*_m
    except Exception: pass
    cm=librosa.feature.chroma_cqt(y=y,sr=sr).mean(axis=1)
    MA=np.array([6.35,2.23,3.48,2.33,4.38,4.09,2.52,5.19,2.39,3.66,2.29,2.88])
    MI=np.array([6.33,2.68,3.52,5.38,2.60,3.53,2.54,4.75,3.98,2.69,3.34,3.17])
    best=(-1,0,'maior')
    for i in range(12):
        for perf,tp in [(MA,'maior'),(MI,'menor')]:
            c=np.corrcoef(cm,np.roll(perf,i))[0,1]
            if c>best[0]: best=(c,i,tp)
    toni,tipo=best[1],best[2]
    graus=[0,2,4,5,7,9,11] if tipo=='maior' else [0,2,3,5,7,8,10]
    escala=set((toni+g)%12 for g in graus)
    _ctx_final={'bpm':bpm,'batida':60.0/bpm,'fase':fase,'toni':toni,'tipo':tipo,'escala':escala,'bpm_manual':bool(bpm_manual)}
    _salvar_ctx(_ctx_final)
    return _ctx_final

# ---- exporta XML com divisions pequeno (Ajuste 1) ----
def _exporta_xml_limpo(stream_obj, caminho):
    stream_obj.write('musicxml', fp=caminho)
    try:
        txt=open(caminho,encoding='utf-8').read()
        m=re.search(r'<divisions>(\d+)</divisions>', txt)
        if m and int(m.group(1))>960:
            div_atual=int(m.group(1)); novo=480; fator=novo/div_atual
            txt=re.sub(r'<duration>(\d+)</duration>', lambda mm:f'<duration>{max(1,round(int(mm.group(1))*fator))}</duration>', txt)
            txt=txt.replace(f'<divisions>{div_atual}</divisions>', f'<divisions>{novo}</divisions>',1)
            open(caminho,'w',encoding='utf-8').write(txt)
    except Exception: pass

# ---- CAMADA 2: motor de partitura de UMA pista (motor v12 INTOCADO) ----
def transcrever_pista(caminho_pista, ctx, base, nome_parte='Voz Principal',
                      clave='sol', modo='registro', ministracao_seg=0,
                      inicio_musica_seg=0.0, usar_grade_banda=True):
    bpm=ctx['bpm']; batida=ctx['batida']; toni=ctx['toni']; tipo=ctx['tipo']; escala=ctx['escala']
    yv,sr=librosa.load(caminho_pista,sr=22050,mono=True)
    # corta ministracao (introducao falada) se informado
    if ministracao_seg and ministracao_seg>0.5:
        yv=yv[int(ministracao_seg*sr):]
    # ---- MOTOR v12 (identico) ----
    o_e=librosa.onset.onset_strength(y=yv,sr=sr,hop_length=256)
    o_s=librosa.onset.onset_strength(y=yv,sr=sr,hop_length=256,feature=librosa.feature.melspectrogram)
    env=(0.3*o_e/(o_e.max()+1e-9)+0.7*o_s/(o_s.max()+1e-9))
    picos,_=find_peaks(env,distance=int(0.06/(256/sr)),prominence=0.1)
    onset_t=librosa.frames_to_time(picos,sr=sr,hop_length=256)
    f0,vf,vp=librosa.pyin(yv,fmin=100,fmax=600,sr=sr,hop_length=256)
    f0_t=librosa.times_like(f0,sr=sr,hop_length=256)
    rms=librosa.feature.rms(y=yv,frame_length=1024,hop_length=256)[0]
    rms_t=librosa.frames_to_time(np.arange(len(rms)),sr=sr,hop_length=256)
    def ee(t): return rms[int(np.argmin(np.abs(rms_t-t)))]
    def pe(a,b):
        mask=(f0_t>=a)&(f0_t<b); vals=f0[mask]; vals=vals[~np.isnan(vals)]
        if len(vals)==0: return None
        mid=np.round(librosa.hz_to_midi(vals)).astype(int); u,c=np.unique(mid,return_counts=True); return int(u[np.argmax(c)])
    def corr(m):
        if NOTAS[m%12] in escala: return m
        for d in [1,-1,2,-2]:
            if NOTAS[(m+d)%12] in escala: return m+d
        return m
    lim_e=rms.max()*0.10
    eventos=[]
    for i,t in enumerate(onset_t):
        if ee(t)<lim_e: continue
        tp=onset_t[i+1] if i+1<len(onset_t) else t+2
        p=pe(t,min(tp,t+0.4))
        if p: eventos.append((corr(p),float(t),float(tp)))
    if not eventos: return None  # pista sem notas (a 7C pula, nao aborta)
    # GRADE DE SEMICOLCHEIA PURA (motor v12)
    GRID=0.25
    fase=ctx.get('fase',0.0)
    # ANCORAGEM:
    # usar_grade_banda = a voz encaixa na GRADE REAL da banda (mapa de tempo mestre).
    # A posicao de cada nota vem da fase+bpm da banda, nao da voz isolada.
    # (ministracao_seg ja foi cortado do audio; some de volta para o tempo absoluto)
    if usar_grade_banda:
        # IMPORTANTE: a ministracao JA foi cortada do audio (yv). Entao o tempo dentro
        # da pista (s0) ja comeca no CANTO. NAO somar ministracao de volta (isso
        # desfaria o corte). A ministracao vira PAUSA no inicio (compassos vazios) no
        # modo ao_vivo, ou e simplesmente ignorada no modo registro.
        if modo=='registro':
            # REGISTRO (estilo Salatiel): a voz comeca no TEMPO 1 do compasso 1.
            t0_abs = ministracao_seg + eventos[0][1]
            desloc = (t0_abs - fase)/batida   # tira tudo antes da 1a nota (inclui ministracao)
        else:  # AO_VIVO: a voz cai onde esta na musica; a ministracao vira PAUSA inicial.
            # posiciona relativo ao inicio real da musica (fase), mantendo o gap da ministracao.
            desloc = 0.0
        qs=[]
        for p,s0,s1 in eventos:
            t_abs = ministracao_seg + s0   # tempo absoluto na musica (canto real)
            b0 = (t_abs - fase)/batida - desloc
            if b0<-0.01: continue
            qs.append([round(b0/GRID)*GRID,p])
    else:
        # modo antigo (ancora na voz isolada) - fallback
        if modo=='ao_vivo':
            offset_pista=(ministracao_seg+eventos[0][1])-inicio_musica_seg
            if offset_pista<0: offset_pista=0
            inicio_local=eventos[0][1]
        else:
            offset_pista=0.0; inicio_local=eventos[0][1]
        qs=[]
        for p,s0,s1 in eventos:
            b0=(s0-inicio_local)/batida + (offset_pista/batida if modo=='ao_vivo' else 0)
            if b0<-0.01: continue
            qs.append([round(b0/GRID)*GRID,p])
    qs2=[]
    for qi,p in qs:
        if qs2 and abs(qs2[-1][0]-qi)<0.01: continue
        qs2.append([qi,p])
    s=stream.Part()
    s.insert(0,m21tempo.MetronomeMark(number=bpm))
    s.insert(0,key.Key(NOTAS[toni].lower() if tipo=='menor' else NOTAS[toni],'minor' if tipo=='menor' else 'major'))
    s.insert(0,meter.TimeSignature('4/4'))
    for k in range(len(qs2)):
        ini_b=qs2[k][0]; fim_b=qs2[k+1][0] if k+1<len(qs2) else ini_b+1
        dur=max(GRID, round((fim_b-ini_b)/GRID)*GRID)
        n=m21note.Note(qs2[k][1]); n.duration=duration.Duration(dur); s.insert(ini_b,n)
    s2=s.makeMeasures(); s2.makeTies(inPlace=True)
    # ---- Bloco 1: metadados ----
    try:
        instr=m21inst.Vocalist() if clave!='fa' else m21inst.Bass()
        instr.partName=nome_parte; s2.insert(0,instr)
    except Exception: pass
    s2.partName=nome_parte
    md=m21meta.Metadata(); md.title=base; md.movementName=base
    try:
        from music21.metadata import Contributor
        md.addContributor(Contributor(role='transcriber', name='Orivox Transcriber - Desenvolvido por Raphael Silveira'))
    except Exception: pass
    s2.insert(0,md)
    primeiro=s2.getElementsByClass('Measure')[0]
    primeiro.insert(0,m21tempo.MetronomeMark(number=round(bpm)))
    # clave
    try:
        nm=[n.pitch.midi for n in s2.recurse().notes]
        if clave=='fa':
            for cl in list(primeiro.getElementsByClass(m21clef.Clef)): primeiro.remove(cl)
            primeiro.insert(0,m21clef.BassClef())
        elif nm and (sum(nm)/len(nm)<64):
            for cl in list(primeiro.getElementsByClass(m21clef.Clef)): primeiro.remove(cl)
            primeiro.insert(0,m21clef.Treble8vbClef())
    except Exception: pass
    # acidentes de precaucao
    try:
        for n in s2.recurse().notes:
            if n.pitch.accidental is not None and n.pitch.accidental.name=='natural':
                n.pitch.accidental.displayStatus=False
    except Exception: pass
    return s2


# ===== PAINEL DE CONTROLE INTERATIVO (clique para escolher) =====
import ipywidgets as widgets
from IPython.display import display, HTML

_modo = widgets.ToggleButtons(
    options=[('  Ao Vivo / Geral  ','ao_vivo'),('  Registro  ','registro')],
    value='registro', button_style='info', description='Modalidade:',
    style={'description_width':'110px'})
_ministracao = widgets.IntSlider(value=0,min=0,max=180,step=1,
    description='Ministracao (s):', continuous_update=False,
    style={'description_width':'130px'}, layout=widgets.Layout(width='520px'))
_bpm = widgets.IntText(value=0, description='BPM (0=auto):',
    style={'description_width':'130px'}, layout=widgets.Layout(width='300px'))
_dica = widgets.HTML(value="<i style='color:#0a8'>Ao Vivo = todas as pistas alinhadas (pausas iniciais). Registro = comeca no 1o compasso.</i>")

display(HTML('''<style>
/* botoes do painel na cor turquesa do Orivox */
.widget-toggle-button { font-weight:600 !important; }
.jupyter-widgets.widget-toggle-button.mod-active,
.widget-toggle-button:checked,
button.mod-active { background:#00c4b4 !important; color:#04231f !important; }
.widget-toggle-buttons .widget-toggle-button.jupyter-button.mod-active { background:#00c4b4 !important; }
.widget-slider .noUi-connect, .ui-slider-range { background:#00c4b4 !important; }
.widget-label { color:#0a8f84 !important; font-weight:600; }
</style>'''))
_bpm_mult = widgets.Dropdown(options=[('x1 (normal)',1.0),('x0.5 (metade)',0.5),('x2 (dobro)',2.0)], value=1.0, description='Multiplicar BPM:', style={'description_width':'130px'})
print('CONFIGURE a partitura (vale para a Celula 7 - voz):')
display(_modo,_ministracao,_bpm,_bpm_mult,_dica)

# funcoes de acesso rapido aos valores escolhidos
def _get_modo(): return _modo.value
def _get_min(): return int(_ministracao.value)
def _get_bpm(): return int(_bpm.value) if int(_bpm.value)>0 else None
def _get_bpm_mult():
    try: return float(_bpm_mult.value)
    except Exception: return 1.0
print('\nPainel pronto. Ajuste acima e siga para a Celula 7 (voz).')
```

```python
#@title 8 - Partitura COMPLETA pelo MuScriptor (pipeline NATIVA, igual ao site) { display-mode: "form" }
# ============================================================================
# ORIVOX ReSound Mixers V27 - Partitura completa pelo MuScriptor
#   Esta celula chama a PIPELINE NATIVA do pacote muscriptor (a MESMA do site
#   muscriptor.kyutai.org): transcreve a mistura completa, quantiza as notas na
#   grade (quantize=True) e engrava com o MuseScore 4 (write_sheets). Resultado
#   identico ao site: XML + full score + 1 PDF por instrumento + tablaturas
#   (detectadas pelo StringData do instrumento: guitarra/baixo sim, cello nao).
#   Em volta, so o que e seu: remapeamento de bateria do Raphael (opcional),
#   nomes versionados "V27 {musica} - {Instrumento}" e MIDI por instrumento.
# Codigo do MuScriptor: MIT | Pesos do modelo: CC-BY-NC-4.0 (uso nao comercial).
# Desenvolvido por Raphael Silveira.
# ============================================================================
import subprocess, sys, os, glob, re, shutil, gc

VERSAO = "V27"

#@markdown **Fonte do audio:** desmarque p/ usar o audio da Celula 2; marque p/ enviar agora:
ENVIAR_AUDIO_AGORA = True  #@param {type:"boolean"}
#@markdown **Tamanho do MuScriptor** (medium = o que o site usa; large pode dropar trechos densos):
TAMANHO = "medium"  #@param ["small", "medium", "large"]
#@markdown **Deteccao de andamento** (forcar = mais confiavel; best-effort cai em 120 se falhar):
ANDAMENTO = "forcar (recomendado)"  #@param ["forcar (recomendado)", "automatico (best-effort)", "desligado (sem grade)"]
#@markdown **Bateria:** desmarcado = bateria NATIVA (igual ao site, padrao); marque p/ usar o SEU remapeamento (regra do Raphael) na principal:
REMAP_BATERIA = False  #@param {type:"boolean"}
#@markdown **Gerar TAMBEM a bateria do outro mapa** (sai um PDF de bateria extra p/ comparar) - padrao ligado:
BATERIA_AMBAS = True  #@param {type:"boolean"}
#@markdown **Baixar tudo num ZIP:**
GERAR_ZIP = True  #@param {type:"boolean"}

OUT = '/content/partitura_v27'
shutil.rmtree(OUT, ignore_errors=True); os.makedirs(OUT, exist_ok=True)

# -------------------- FONTE DO AUDIO --------------------
if ENVIAR_AUDIO_AGORA:
    from google.colab import files
    print('Envie o audio COMPLETO (mp3/wav):')
    env = files.upload(); nome = list(env.keys())[0]; cam = '/content/' + nome
    if os.path.exists(nome) and os.path.abspath(nome) != cam: os.replace(nome, cam)
else:
    cands = glob.glob('/content/entrada/*')
    if not cands:
        raise SystemExit('Nao achei audio em /content/entrada. Rode a Celula 2 ou marque "enviar agora".')
    cam = cands[0]
base = os.path.splitext(os.path.basename(cam))[0]
print('Audio:', os.path.basename(cam))

# -------------------- INSTALL (pacote + MuseScore 4) --------------------
# ATENCAO: a pipeline de partitura (write_sheets + quantize) existe no repositorio
# do GitHub, NAO na versao 0.3.0 publicada no PyPI. Por isso instalamos do GitHub,
# fixado no commit testado. Codigo MIT (instalar da fonte e permitido).
print('Instalando MuScriptor (do GitHub, versao com a pipeline de partitura)... (1a vez ~2-3 min)')
MUSC_GIT = 'git+https://github.com/muscriptor/muscriptor.git@7f213afecf23bd6a1b8672aa223690ee9807cefb'
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', MUSC_GIT, 'librosa', 'soundfile', 'mido'], check=False)

def instalar_musescore4():
    """Instala o MuseScore 4 (AppImage do GitHub) e devolve o caminho do AppRun.
    write_sheets EXIGE MuseScore 4+: o 3.6 nao carrega a afinacao das cordas
    e nao gera tablatura. A URL antiga de /download devolve HTML (403), por isso
    o AppImage vem direto do release do GitHub, que e um binario de verdade."""
    extr = '/content/squashfs-root'; apprun = f'{extr}/AppRun'
    if os.path.exists(apprun): return apprun
    libs = ['libfuse2', 'libxcb-xinerama0', 'libxcb-icccm4', 'libxcb-image0', 'libxcb-keysyms1',
            'libxcb-randr0', 'libxcb-render-util0', 'libxcb-shape0', 'libxkbcommon-x11-0',
            'libegl1', 'libgl1', 'libxcb-cursor0', 'libnss3', 'libxcomposite1', 'libxdamage1',
            'libasound2t64', 'libodbc2']
    subprocess.run(['apt-get', '-qq', 'install', '-y'] + libs, check=False)
    urls = ['https://github.com/musescore/MuseScore/releases/download/v4.5.2/MuseScore-Studio-4.5.2.251141401-x86_64.AppImage',
            'https://github.com/musescore/MuseScore/releases/download/v4.4.4/MuseScore-Studio-4.4.4.243461245-x86_64.AppImage']
    app = '/content/MuseScore4.AppImage'
    for url in urls:
        subprocess.run(['wget', '-q', '-O', app, url], check=False)
        ok = os.path.exists(app) and os.path.getsize(app) > 50_000_000
        if ok:
            with open(app, 'rb') as fh: ok = fh.read(4) == b'\x7fELF'
        if not ok: continue
        os.chmod(app, 0o755); cwd = os.getcwd(); os.chdir('/content')
        subprocess.run([app, '--appimage-extract'], check=False, capture_output=True); os.chdir(cwd)
        if os.path.exists(apprun):
            print('MuseScore 4 pronto:', url.split('/')[-1]); return apprun
    return None

MSCORE = instalar_musescore4()
if not MSCORE:
    raise SystemExit(
        'ATENCAO: o MuseScore 4 NAO instalou (sem ele nao da para engravar igual ao site).\n'
        'A partitura limpa (quantizada, com tablatura) EXIGE MuseScore 4+. O 3.6 nao serve.\n'
        'Verifique a conexao do Colab com o GitHub e rode a celula de novo.')
os.environ['MUSCRIPTOR_MUSESCORE'] = MSCORE

import librosa, soundfile as sf
y, sr = librosa.load(cam, sr=None, mono=False)
wavpath = '/content/_mus_full.wav'
sf.write(wavpath, y.T if getattr(y, 'ndim', 1) > 1 else y, sr)

# -------------------- MUSCRIPTOR (transcricao NATIVA + quantizacao) --------------------
ESPELHO = f'hf://cocktailpeanut/muscriptor-{TAMANHO}/model.safetensors'  # espelho aberto (sem token)
MODO = {'forcar (recomendado)': True, 'automatico (best-effort)': 'best-effort', 'desligado (sem grade)': False}[ANDAMENTO]
print(f'MuScriptor ({TAMANHO}) no audio completo, andamento={MODO}, quantize=True...')
from muscriptor import TranscriptionModel
try:
    model = TranscriptionModel.load_model(ESPELHO)
    try:
        midi_bytes, grid = model.transcribe_and_postprocess(wavpath, detect_tempo=MODO, quantize=True)
    except Exception as e_bt:
        if MODO is True:
            print(f'"forcar" nao achou andamento fixo ({e_bt}); tentando best-effort...')
            midi_bytes, grid = model.transcribe_and_postprocess(wavpath, detect_tempo='best-effort', quantize=True)
        else:
            raise
    print('MuScriptor concluido.')
except Exception as e:
    print('ERRO no MuScriptor:', e); raise SystemExit('Verifique a GPU (T4) e a conexao.')
finally:
    # libera a GPU para a Celula 3 (Demucs) nao dar "CUDA busy" depois desta.
    try:
        del model
    except Exception:
        pass
    gc.collect()
    try:
        import torch
        if torch.cuda.is_available(): torch.cuda.empty_cache()
    except Exception:
        pass

QUANT = grid is not None and getattr(grid, 'beat_subdivision', None) is not None

# -------------------- REMAP DE BATERIA (regra do Raphael) --------------------
import mido, io
REMAP = {35:"bumbo",36:"bumbo",37:"caixa_aro",38:"caixa",39:"caixa",40:"caixa",41:"surdo",43:"surdo",
         42:"hihat_fechado",44:"hihat_semi",46:"hihat_aberto",45:"tom2",47:"tom2",48:"tom1",50:"tom1",65:"tom1",
         49:"crash",52:"crash",54:"hihat_fechado",57:"crash",55:"crash",51:"ride",53:"ride",59:"ride"}
PECA_TECLA = {"bumbo":36,"caixa":38,"caixa_aro":37,"surdo":41,"hihat_fechado":42,"hihat_semi":44,
              "hihat_aberto":46,"tom2":45,"tom1":48,"crash":49,"ride":51}
def _is_drum(tr): return any(m.type=='note_on' and m.velocity>0 and getattr(m,'channel',None)==9 for m in tr)
def remapeia_bytes(midi_in_bytes):
    mid = mido.MidiFile(file=io.BytesIO(midi_in_bytes))
    tempo = next((x.tempo for tr in mid.tracks for x in tr if x.type=='set_tempo'), 500000)
    ts = next(((x.numerator,x.denominator) for tr in mid.tracks for x in tr if x.type=='time_signature'), None)
    out = mido.MidiFile(ticks_per_beat=mid.ticks_per_beat, type=1)
    for tr in mid.tracks:
        if not _is_drum(tr): out.tracks.append(tr); continue
        nome = next((m.name for m in tr if m.type=='track_name'), None); ons=[]; ab=0
        for m in tr:
            ab += m.time
            if m.type=='note_on' and m.velocity>0 and getattr(m,'channel',None)==9:
                pc = REMAP.get(m.note)
                if pc: ons.append((ab, PECA_TECLA[pc]))
        ev=[]
        for tk,p in ons: ev += [(tk,1,'on',p),(tk+20,0,'off',p)]
        ev.sort(key=lambda e:(e[0],e[1])); nt = mido.MidiTrack()
        if nome: nt.append(mido.MetaMessage('track_name', name=nome, time=0))
        nt.append(mido.MetaMessage('set_tempo', tempo=tempo, time=0))
        if ts: nt.append(mido.MetaMessage('time_signature', numerator=ts[0], denominator=ts[1], time=0))
        last=0
        for tk,_o,k,p in ev:
            d=max(0,tk-last); last=tk
            nt.append(mido.Message('note_on' if k=='on' else 'note_off', note=p, velocity=100 if k=='on' else 0, channel=9, time=d))
        out.tracks.append(nt)
    buf=io.BytesIO(); out.save(file=buf); return buf.getvalue()

midi_principal = remapeia_bytes(midi_bytes) if REMAP_BATERIA else midi_bytes
if REMAP_BATERIA: print('Remapeamento de bateria aplicado (regra do Raphael).')

# -------------------- ENGRAVE (write_sheets = pipeline do site) --------------------
from pathlib import Path
from muscriptor.utils.sheets import write_sheets
raw = Path(OUT) / '_sheets'
print('Engravando com o MuseScore 4 (write_sheets, igual ao site)...')
gerados = [Path(p) for p in write_sheets(midi_principal, raw, quantized=QUANT)]

# bateria do OUTRO mapa, se pedida (um PDF extra de bateria para comparar)
bateria_extra = None
if BATERIA_AMBAS:
    outro = midi_bytes if REMAP_BATERIA else remapeia_bytes(midi_bytes)
    raw2 = Path(OUT) / '_sheets_outra_bateria'
    try:
        extra = [Path(p) for p in write_sheets(outro, raw2, quantized=QUANT)]
        bateria_extra = next((p for p in extra if 'drum' in p.name.lower() and p.suffix=='.pdf'), None)
    except Exception as e:
        print('  (bateria extra falhou:', e, ')')

# -------------------- NOMES VERSIONADOS + MIDI POR INSTRUMENTO --------------------
def instrumento_amigavel(nome):
    n = nome.lower()
    if 'piano' in n: return 'Piano'
    if 'drum' in n: return 'Drums'
    if 'bass' in n or 'baixo' in n: return 'Bass'
    if 'distorted' in n and 'guitar' in n: return 'Guitar (distorted)'
    if 'clean' in n and 'guitar' in n: return 'Guitar (clean)'
    if ('acoustic' in n or 'classical' in n) and 'guitar' in n: return 'Guitar (acoustic)'
    if 'guitar' in n: return 'Guitar'
    if 'voice' in n or 'tenor' in n or 'soprano' in n or 'vocal' in n: return 'Voice'
    if 'viol' in n or 'string' in n or 'cello' in n: return 'Strings'
    if 'synth' in n: return 'Synth'
    if 'brass' in n or 'trumpet' in n or 'trombone' in n or 'sax' in n: return 'Brass'
    if 'flute' in n or 'clarinet' in n or 'oboe' in n: return 'Winds'
    return re.sub(r'[^A-Za-z0-9 ]+', '', nome).strip().title() or 'Part'

def destino(nome_amigavel, tipo):
    # tipo: 'notacao' | 'tab' | 'score' | 'xml' | 'midi'
    if tipo == 'score':  return f'{VERSAO} {base} - Partitura completa.pdf'
    if tipo == 'xml':    return f'{VERSAO} {base} - score.musicxml'
    if tipo == 'tab':    return f'{VERSAO} {base} - {nome_amigavel} (tab).pdf'
    if tipo == 'midi':   return f'{VERSAO} {base} - {nome_amigavel}.mid'
    return f'{VERSAO} {base} - {nome_amigavel}.pdf'

finais = []
for p in gerados:
    nm = p.name
    if nm == 'full_score.pdf':
        dst = destino(None, 'score')
    elif nm == 'score.musicxml':
        dst = destino(None, 'xml')
    elif nm == 'score.mid':
        dst = f'{VERSAO} {base} - Geral.mid'   # MIDI geral (alem dos por instrumento)
    elif nm.endswith('_tab.pdf'):
        bruto = re.sub(r'^\d+_', '', nm[:-8]).replace('_', ' ')
        dst = destino(instrumento_amigavel(bruto), 'tab')
    elif nm.endswith('.pdf'):
        bruto = re.sub(r'^\d+_', '', nm[:-4]).replace('_', ' ')
        dst = destino(instrumento_amigavel(bruto), 'notacao')
    else:
        dst = nm
    final = Path(OUT) / dst
    shutil.copyfile(p, final); finais.append(final)

if bateria_extra is not None:
    etq = 'nativa' if REMAP_BATERIA else 'remap'
    final = Path(OUT) / f'{VERSAO} {base} - Drums ({etq}).pdf'
    shutil.copyfile(bateria_extra, final); finais.append(final)

# MIDI por instrumento (split do score.mid engravado, mantendo tempo/compasso)
score_mid = next((p for p in gerados if p.name == 'score.mid'), None)
if score_mid is not None:
    mid = mido.MidiFile(str(score_mid)); meta = mid.tracks[0] if mid.tracks else None
    for tr in mid.tracks:
        nome = next((m.name for m in tr if m.type == 'track_name'), None)
        tem_nota = any(m.type == 'note_on' and m.velocity > 0 for m in tr)
        if not (nome and tem_nota): continue
        um = mido.MidiFile(ticks_per_beat=mid.ticks_per_beat, type=1)
        if meta is not None and meta is not tr: um.tracks.append(meta)
        um.tracks.append(tr)
        final = Path(OUT) / destino(instrumento_amigavel(nome), 'midi')
        um.save(str(final)); finais.append(final)

# -------------------- ENTREGA --------------------
print(f'\nPronto! {len(finais)} arquivo(s). Desenvolvido por Raphael Silveira.')
for f in sorted(finais): print('  ', f.name)
from google.colab import files as _f
if GERAR_ZIP:
    import zipfile
    zp = Path(OUT) / f'{VERSAO} - {base}_partituras.zip'
    with zipfile.ZipFile(zp, 'w', zipfile.ZIP_DEFLATED) as z:
        for f in finais:
            if os.path.exists(f): z.write(f, os.path.basename(f))
    print('ZIP:', zp.name)
    try: _f.download(str(zp))
    except Exception: pass
else:
    for f in finais:
        try: _f.download(str(f))
        except Exception: pass
```

```python
#@title 9 - CIFRA profissional (motor neural + PDF estilo Orivox) { display-mode: "form" }
# Desenvolvido por Raphael Silveira.
# Motor: basic-pitch (neural, GPU) nas notas da HARMONIA separada -> acordes por compasso.
#        Fallback automatico: croma CQT (librosa) se o basic-pitch nao estiver disponivel.
# Letra: openai-whisper (tempos por palavra) para acordes sobre a letra (estilo Moises).
# PDF:   layout profissional com a marca Orivox e "Chords generated with Orivox".
import os, glob, subprocess, sys, importlib
import numpy as np

#@markdown **Enviar audio agora** (desmarque p/ usar o da Celula 2):
ENVIAR_AUDIO_AGORA = True  #@param {type:"boolean"}
#@markdown **Usar as pistas separadas** (mais limpo; precisa ter rodado a Celula 3):
USAR_STEMS = False  #@param {type:"boolean"}
#@markdown **Acordes sobre a letra** (transcreve a letra; desmarque p/ so a grade):
COM_LETRA = True  #@param {type:"boolean"}

# ---- instala cada coisa SEPARADAMENTE (uma falha nao derruba o resto) ----
def ter(pkg, mod=None):
    mod = mod or pkg.replace('-','_')
    try: return importlib.import_module(mod)
    except Exception: pass
    print('  instalando %s...' % pkg)
    subprocess.run([sys.executable,'-m','pip','install','-q',pkg])
    try: return importlib.import_module(mod)
    except Exception as e:
        print('  (nao instalou %s: %s)' % (pkg, e)); return None
print('Preparando ferramentas (instala so o que faltar)...')
ter('librosa'); ter('soundfile'); ter('reportlab')
import librosa, soundfile as sf

# ---- fonte de audio ----
if ENVIAR_AUDIO_AGORA:
    from google.colab import files
    print('Envie o audio:'); env=files.upload(); nome=list(env.keys())[0]; CAM='/content/'+nome
    if os.path.exists(nome) and os.path.abspath(nome)!=CAM: os.replace(nome,CAM)
else:
    cands=glob.glob('/content/entrada/*')
    if not cands: raise SystemExit('Nao achei audio. Rode a Celula 2 ou marque "enviar agora".')
    CAM=cands[0]
BASE=os.path.splitext(os.path.basename(CAM))[0]
print('Audio:', os.path.basename(CAM))

# ---- HARMONIA: mistura stems (sem bateria/voz) se houver ----
HARM=CAM; usei_stems=False
if USAR_STEMS:
    alvos=[]
    for nome in ['bass','other','guitar','piano']:
        g=glob.glob('/content/saida/*/*/'+nome+'.*')
        if g: alvos.append(g[0])
    if alvos:
        sigs=[]; srr=22050
        for a in alvos:
            yy,srr=librosa.load(a,sr=22050,mono=True); sigs.append(yy)
        L=max(len(s) for s in sigs); mix=np.zeros(L)
        for s in sigs: mix[:len(s)]+=s
        mix=mix/(np.max(np.abs(mix))+1e-9)
        HARM='/content/_harmonia.wav'; sf.write(HARM, mix, srr); usei_stems=True
        print('Analisando a HARMONIA de %d pistas separadas (sem bateria/voz).' % len(alvos))
if not usei_stems:
    print('Analisando a musica completa (rode a Celula 3 p/ separar e ficar mais limpo).')

# ---- tom, BPM e beats (librosa) ----
NOTAS=['C','C#','D','D#','E','F','F#','G','G#','A','A#','B']
yfull,sr=librosa.load(CAM,sr=22050,mono=True); dur=len(yfull)/sr
try:
    tempo,beatf=librosa.beat.beat_track(y=yfull,sr=sr,trim=False)
    beats=librosa.frames_to_time(beatf,sr=sr); bpm=int(round(float(np.atleast_1d(tempo)[0])))
except Exception:
    bpm=100; beats=np.arange(0,dur,60/bpm)
cmG=librosa.feature.chroma_cqt(y=yfull,sr=sr).mean(axis=1)
MA=np.array([6.35,2.23,3.48,2.33,4.38,4.09,2.52,5.19,2.39,3.66,2.29,2.88]); MI=np.array([6.33,2.68,3.52,5.38,2.60,3.53,2.54,4.75,3.98,2.69,3.34,3.17])
bk=(-9,0,True)
for i in range(12):
    for p,mj in [(MA,True),(MI,False)]:
        cc=float(np.corrcoef(np.roll(cmG,-i),p)[0,1])
        if cc>bk[0]: bk=(cc,i,mj)
root,maj=bk[1],bk[2]
# grafia: bemois para tons com bemol (ex. F menor -> Bb/Db/Ab), sustenidos p/ os de sustenido
FLATS=['C','Db','D','Eb','E','F','Gb','G','Ab','A','Bb','B']
usa_flat = (root not in {7,2,9,4,11,6}) if maj else (root not in {4,11,6,1,8})
NOMES = FLATS if usa_flat else NOTAS
tomtxt=NOMES[root]+(' maior' if maj else ' menor')
graus=[0,2,4,5,7,9,11] if maj else [0,2,3,5,7,8,10]; escala=set((root+x)%12 for x in graus)
QUAL=[('',[0,4,7]),('m',[0,3,7]),('maj7',[0,4,7,11]),('m7',[0,3,7,10]),('7',[0,4,7,10])]
MOLDS=[]
for r in range(12):
    for suf,ii in QUAL:
        v=np.zeros(12)
        for it in ii: v[(r+it)%12]=1
        v/=np.sqrt(len(ii)); MOLDS.append((r,NOMES[r]+suf,v))
def rotular(vec12, prev):
    n=np.linalg.norm(vec12)+1e-9; m=vec12/n; bv=-9; bb=MOLDS[0]
    for r,l,v in MOLDS:
        d=float(np.dot(m,v))
        if r in escala: d+=0.08
        if r==prev: d+=0.04
        if d>bv: bv=d; bb=(r,l,v)
    return bb

# ---- MOTOR: basic-pitch (neural) nas notas; fallback croma CQT ----
notes=None
if ter('basic-pitch','basic_pitch'):
    try:
        from basic_pitch.inference import predict
        print('Detectando notas com basic-pitch (neural)...')
        _,_,notes=predict(HARM)
        print('basic-pitch: %d notas.' % len(notes))
    except Exception as e:
        print('basic-pitch falhou (%s). Usando croma CQT (confiavel).' % e)
else:
    print('basic-pitch indisponivel. Usando croma CQT (confiavel).')

if notes:
    def vec_compasso(t0,t1):
        w=np.zeros(12)
        for ev in notes:
            s,e,p = ev[0],ev[1],ev[2]; a=ev[3] if len(ev)>3 else 1.0
            ov=max(0.0,min(e,t1)-max(s,t0))
            if ov>0: w[int(p)%12]+=ov*(a or 1.0)
        return w if w.sum()>1e-9 else None
else:
    Ccqt=librosa.feature.chroma_cqt(y=librosa.load(HARM,sr=22050,mono=True)[0],sr=22050)
    tcq=librosa.frames_to_time(np.arange(Ccqt.shape[1]),sr=22050)
    csq=Ccqt/(np.linalg.norm(Ccqt,axis=0,keepdims=True)+1e-9)
    def vec_compasso(t0,t1):
        i0=np.searchsorted(tcq,t0); i1=np.searchsorted(tcq,t1)
        if i1<=i0: return None
        return np.median(csq[:,i0:i1],axis=1)

prog=[]; prev=None
for k in range(0,max(1,len(beats)-1),4):
    t0=beats[k]; t1=beats[k+4] if k+4<len(beats) else dur
    vv=vec_compasso(t0,t1)
    if vv is None:
        if prog: prog.append((t0,t1,prog[-1][2]))
        continue
    r,l,_=rotular(vv,prev); prog.append((t0,t1,l)); prev=r
print('Tom %s | %d BPM | %d compassos' % (tomtxt,bpm,len(prog)))
def acorde_em(t):
    best=None
    for (s,e,l) in prog:
        if s<=t+1e-6: best=l
        else: break
    return best

# ---- LETRA (openai-whisper, opcional) ----
segs=[]
if COM_LETRA and ter('openai-whisper','whisper'):
    try:
        import whisper, torch
        dev='cuda' if torch.cuda.is_available() else 'cpu'
        print('Transcrevendo a letra (Whisper, %s)...' % dev)
        wm=whisper.load_model('small', device=dev)
        voz=glob.glob('/content/vozes/*(Vocals)*') or glob.glob('/content/saida/*/*/vocals.*') or [CAM]
        r=wm.transcribe(voz[0], language='pt', word_timestamps=True, fp16=(dev=='cuda'))
        for s in r.get('segments',[]):
            ws=[(float(w['start']), str(w['word']).strip()) for w in s.get('words',[]) if str(w.get('word','')).strip()]
            segs.append({'text':s['text'].strip(),'start':float(s['start']),'words':ws})
        print('Letra: %d linhas.' % len(segs))
    except Exception as e:
        print('Sem letra (segue so com acordes):', e)

# ---- PDF profissional (marca Orivox) ----
from reportlab.lib.pagesizes import A4
from reportlab.pdfgen import canvas
from reportlab.lib.units import cm as CM
import textwrap
PDF='/content/Orivox - %s.pdf' % BASE
c=canvas.Canvas(PDF, pagesize=A4); W,H=A4; TEAL=(0.164,0.831,0.784)
def rodape():
    c.setFont('Helvetica',8); c.setFillColorRGB(0.6,0.6,0.6); c.drawCentredString(W/2,0.8*CM,'Chords generated with Orivox')
def cabecalho():
    c.setFont('Times-Bold',22); c.setFillColorRGB(0.10,0.10,0.12); c.drawString(2*CM,H-2*CM,'Oriv')
    w=c.stringWidth('Oriv','Times-Bold',22); c.setFillColorRGB(*TEAL); c.drawString(2*CM+w,H-2*CM,'ox')
    c.setFillColorRGB(0.10,0.10,0.12); c.setFont('Helvetica-Bold',13); c.drawString(2*CM,H-2.9*CM,BASE[:70])
    c.setFont('Helvetica',10); c.setFillColorRGB(0.4,0.4,0.4)
    c.drawString(2*CM,H-3.5*CM,'Transcricao por Orivox  ·  Tom %s  ·  %d BPM  ·  4/4' % (tomtxt,bpm))
    c.setStrokeColorRGB(*TEAL); c.setLineWidth(1.5); c.line(2*CM,H-3.8*CM,W-2*CM,H-3.8*CM)
cabecalho(); rodape(); yv=[H-4.7*CM]; x0=2*CM
def quebra():
    c.showPage(); cabecalho(); rodape(); yv[0]=H-4.7*CM
if segs:
    for seg in segs:
        if yv[0]<2.5*CM: quebra()
        words=seg['words'] or [(seg['start'], seg['text'])]
        c.setFont('Helvetica-Bold',11); c.setFillColorRGB(*TEAL); last=None; desenhou=False; lastx=-1e9
        for i,(wt,wd) in enumerate(words):
            ch=acorde_em(wt)
            if ch and ch!=last:
                prefix=' '.join(w for _,w in words[:i]); px=x0+c.stringWidth(prefix+(' ' if prefix else ''),'Helvetica',11)
                if px<lastx: px=lastx
                if px<W-2*CM:
                    c.drawString(px,yv[0],ch); desenhou=True
                    lastx=px+c.stringWidth(ch,'Helvetica-Bold',11)+6
                last=ch
        if desenhou: yv[0]-=0.45*CM
        c.setFont('Helvetica',11); c.setFillColorRGB(0.10,0.10,0.12); c.drawString(x0,yv[0],seg['text'][:110]); yv[0]-=0.8*CM
else:
    c.setFont('Courier-Bold',11); c.setFillColorRGB(*TEAL)
    labels=[l for _,_,l in prog]
    for k in range(0,len(labels),4):
        if yv[0]<2.5*CM: quebra()
        c.drawString(x0,yv[0],'| '+' | '.join(labels[k:k+4])+' |'); yv[0]-=0.6*CM
c.save()
from google.colab import files as _f
print('PDF gerado:', os.path.basename(PDF)); _f.download(PDF)
```
